17) Create a column that categorizes movies by rating:

a) Rating < 6: Poor

b) 6 ≤ Rating < 7.5: Average

c) Rating ≥ 7.5: Excellent


```python
def category(rating):
    if rating < 6:
        return "Poor"
    elif rating < 7.5:
        return "Average"
    else:
        return "Excellent"

df["Rating_Category"] = df["Rating"].apply(category)

print(df[["Movie_Title","Rating","Rating_Category"]].head())
```
