```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder, LabelEncoder
```

```python
df = pd.read_csv("House_Pricing.csv")

df.head()
```

```python
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate Rows:", duplicate_rows)

df = df.drop_duplicates()

print("Shape after removing duplicate rows:", df.shape)
```

```python
duplicate_columns = []

for i in range(df.shape[1]):
    col1 = df.iloc[:, i]

    for j in range(i + 1, df.shape[1]):
        col2 = df.iloc[:, j]

        if col1.equals(col2):
            duplicate_columns.append(df.columns[j])

print("Duplicate Columns:", duplicate_columns)

df = df.drop(columns=duplicate_columns)

print(df.shape)
```

```python
df.isnull().sum()
```

```python
#Missing values Handling
num_cols = df.select_dtypes(include=['int64','float64']).columns

cat_cols = df.select_dtypes(include=['object']).columns
```

```python
numerical_cols = df.select_dtypes(include=np.number).columns

categorical_cols = df.select_dtypes(include='object').columns

print(numerical_cols)

print(categorical_cols)
```

```python
for col in numerical_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
Q1 = df[scale_columns].quantile(0.25)

Q3 = df[scale_columns].quantile(0.75)

IQR = Q3 - Q1
```

```python
df = df[~((df[scale_columns] < (Q1 - 1.5 * IQR)) |
          (df[scale_columns] > (Q3 + 1.5 * IQR))).any(axis=1)]
```
