```python
import pandas as pd
import numpy as np
```

1. Load the Titanic dataset and display the first 10 rows

```python
df = pd.read_csv("titanic_dataset.csv")

print("First 10 Rows:")
print(df.head(10))
```

2. Find the number of missing values in each column.


```python
print("\nMissing Values:")
print(df.isnull().sum())
```

3. Find the percentage of passengers who survived, grouped by gender

```python
survival_percentage = df.groupby("Sex")["Survived"].mean() * 100

print("\nSurvival Percentage by Gender:")
print(survival_percentage)
```

4. Find the average, median, and standard deviation of passenger ages.


```python
print("\nAge Statistics:")
print("Average Age =", df["Age"].mean())
print("Median Age =", df["Age"].median())
print("Standard Deviation =", df["Age"].std())
```

5. Count how many passengers were in each class (pclass).


```python
print("\nPassengers in each Class:")
print(df["Pclass"].value_counts())
```

6. Show all passengers who were female, under age 18, and survived.


```python
female_survived = df[
    (df["Sex"] == "female") &
    (df["Age"] < 18) &
    (df["Survived"] == 1)
]

print("\nFemale Passengers Under 18 Who Survived:")
print(female_survived)
```

7. Replace missing age values with the median age.


```python
median_age = df["Age"].median()

df["Age"] = df["Age"].fillna(median_age)

print("\nMissing Age Values After Filling:")
print(df["Age"].isnull().sum())
```

8. Create a column age_group with values:

a. 'Child' for age < 18

b. 'Adult' for age 18–60

c. 'Senior' for age > 60

d. Use: apply() or np.where() or custom function


```python
def age_group(age):
    if age < 18:
        return "Child"
    elif age <= 60:
        return "Adult"
    else:
        return "Senior"

df["age_group"] = df["Age"].apply(age_group)

print("\nAge Group:")
print(df[["Age", "age_group"]].head())
```

9. Display the top 5 passengers who paid the highest fare.


```python
top5_fare = df.nlargest(5, "Fare")

print("\nTop 5 Highest Fare Passengers:")
print(top5_fare[["Name", "Fare"]])
```

10. Find the average fare paid, grouped by both pclass and sex.


```python
avg_fare = df.groupby(["Pclass", "Sex"])["Fare"].mean()

print("\nAverage Fare by Class and Gender:")
print(avg_fare)
```

11. Rename the following columns:

a. 'sex' → 'gender'

b. 'fare' → 'ticket_fare'


```python
df.rename(columns={
    "sex": "gender",
    "fare": "ticket_fare"
}, inplace=True)

print(df.columns)
```

12. Count the number of passengers who survived and didn’t survive, grouped by class.


```python
survival_count = df.groupby("Pclass")["Survived"].value_counts()

print(survival_count)
```

13. Find the average age of passengers grouped by both class and gender.

```python
avg_age = df.groupby(
    ["Pclass", "Sex"]
)["Age"].mean()

print(avg_age)
```
