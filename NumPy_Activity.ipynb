1. Create a NumPy array with values from 10 to 49 (inclusive).
● Reshape it into a 4x10 matrix.
● Then extract the last column from the reshaped matrix

```python
import numpy as np

# Create array from 10 to 49
a = np.arange(10, 50)

# Reshape into 4x10 matrix
b = a.reshape(4, 10)

# Extract last column
last_column = b[:, -1]

print("Original Array:")
print(a)

print("\n4x10 Matrix:")
print(b)

print("\nLast Column:")
print(last_column)
```

2. Create two 3×3 matrices:
One filled with all 5s , One with values from 1 to 9
● Add the two matrices
● Multiply them element-wise
● Then find the transpose of the result


```python
import numpy as np

# Matrix filled with 5s
A = np.full((3,3), 5)

# Matrix with values 1 to 9
B = np.arange(1,10).reshape(3,3)

# Add matrices
add_result = A + B

# Element-wise multiplication
mul_result = A * B

# Transpose of multiplication result
transpose_result = mul_result.T

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nAddition:")
print(add_result)

print("\nElement-wise Multiplication:")
print(mul_result)

print("\nTranspose of Multiplication Result:")
print(transpose_result)
```

3. Create a NumPy array of 15 random integers between 10 and 100.
● Replace all values greater than 50 with -1
● Count how many values were replaced.


```python
import numpy as np

# Create an array of 15 random integers between 10 and 100
a = np.random.randint(10, 101, 15)

print("Original Array:")
print(a)

# Count values greater than 50
count = np.sum(a > 50)

# Replace values greater than 50 with -1
a[a > 50] = -1

print("\nModified Array:")
print(a)

print("\nNumber of values replaced:", count)
```

```python
import numpy as np

# Generate a 5x4 array with random float values between 0 and 1
a = np.random.rand(5, 4)

print("Original Array:")
print(a)

# Find row sums
row_sums = np.sum(a, axis=1)

# Find row with maximum sum
max_row_index = np.argmax(row_sums)

print("\nRow Sums:")
print(row_sums)

print("\nRow with Maximum Sum (Index):", max_row_index)
print(a[max_row_index])

# Normalize each row
normalized = (a - a.min(axis=1, keepdims=True)) / \
             (a.max(axis=1, keepdims=True) - a.min(axis=1, keepdims=True))

print("\nNormalized Array:")
print(normalized)
```

5. arr = np.array([1, 3, 5, 3, 7, 9, 1, 5, 5])

● Find all unique elements

● Find the duplicates (elements that appear more than once)


```python
import numpy as np

arr = np.array([1, 3, 5, 3, 7, 9, 1, 5, 5])

# Find unique elements
unique_elements = np.unique(arr)

# Find duplicates
duplicates = unique_elements[np.array([np.sum(arr == x) > 1 for x in unique_elements])]

print("Unique Elements:")
print(unique_elements)

print("\nDuplicate Elements:")
print(duplicates)
```

6. Create a 6×6 NumPy array with values from 1 to 36.

● Extract the middle 4×4 submatrix.

● Then extract all even numbers from this submatrix

```python
import numpy as np

# Create a 6x6 array with values from 1 to 36
arr = np.arange(1, 37).reshape(6, 6)

print("Original Array:")
print(arr)

# Extract the middle 4x4 submatrix
submatrix = arr[1:5, 1:5]

print("\nMiddle 4x4 Submatrix:")
print(submatrix)

# Extract even numbers from the submatrix
even_numbers = submatrix[submatrix % 2 == 0]

print("\nEven Numbers:")
print(even_numbers)
```
